```python
df = pd.read_csv('D:/Project-04/data/cleaned/ev_feature_engineered.csv')

print("Cleaned Dataset Loaded")
```

```python
# Weather vs range utilization

weather_range = df.groupby(
    'weather_condition'
)['range_utilization_pct'].mean().reset_index()

weather_range = weather_range.sort_values(
    by='range_utilization_pct',
    ascending=False
)

ordered_weather = weather_range['weather_condition']

plt.figure(figsize=(10,5))

sns.barplot(
    data=weather_range,
    x='weather_condition',
    y='range_utilization_pct',
    order=ordered_weather
)

plt.title("Average Range Utilization by Weather Condition")

plt.xlabel("Weather Condition")

plt.ylabel("Range Utilization (%)")

plt.xticks(rotation=45)

plt.show()
```

```python
# Cost per KM by Vehicle Type

vehicle_cost = df.groupby(
    'vehicle_type'
)['cost_per_km'].mean().reset_index()

# Sort values

vehicle_cost = vehicle_cost.sort_values(
    by='cost_per_km',
    ascending=False
)

# Ordered categories

ordered_vehicle = vehicle_cost['vehicle_type']

# Plot

plt.figure(figsize=(10,5))

sns.barplot(
    data=vehicle_cost,
    x='vehicle_type',
    y='cost_per_km',
    order=ordered_vehicle
)

plt.title("Average Cost per KM by Vehicle Type")

plt.xlabel("Vehicle Type")

plt.ylabel("Average Cost per KM")

plt.show()
```

```python
# Top cities by energy consumption

city_energy = df.groupby(
    'city'
)['energy_consumed_kwh'].mean().reset_index()

# Sort descending

city_energy = city_energy.sort_values(
    by='energy_consumed_kwh',
    ascending=False
)

# Ordered cities

ordered_cities = city_energy['city']

# Plot

plt.figure(figsize=(12,5))

sns.barplot(
    data=city_energy,
    x='city',
    y='energy_consumed_kwh',
    order=ordered_cities
)

plt.title("Average Energy Consumption by City")

plt.xlabel("City")

plt.ylabel("Average Energy Consumption (kWh)")

plt.xticks(rotation=45)

plt.show()
```

```python
# Charging type vs monthly energy cost

charging_cost = df.groupby(
    'charging_type'
)['monthly_energy_cost'].mean().reset_index()

# Sort descending

charging_cost = charging_cost.sort_values(
    by='monthly_energy_cost',
    ascending=False
)

# Ordered charging types

ordered_charging = charging_cost['charging_type']

# Plot

plt.figure(figsize=(10,5))

sns.barplot(
    data=charging_cost,
    x='charging_type',
    y='monthly_energy_cost',
    order=ordered_charging
)

plt.title("Average Monthly Energy Cost by Charging Type")

plt.xlabel("Charging Type")

plt.ylabel("Monthly Energy Cost")

plt.show()
```

```python
# Range anxiety vs charging station distance

anxiety_distance = df.groupby(
    'range_anxiety_risk'
)['charging_station_distance_km'].mean().reset_index()

# Sort descending

anxiety_distance = anxiety_distance.sort_values(
    by='charging_station_distance_km',
    ascending=False
)

# Ordered categories

ordered_anxiety = anxiety_distance['range_anxiety_risk']

# Plot

plt.figure(figsize=(10,5))

sns.barplot(
    data=anxiety_distance,
    x='range_anxiety_risk',
    y='charging_station_distance_km',
    order=ordered_anxiety
)

plt.title("Charging Station Distance by Range Anxiety Risk")

plt.xlabel("Range Anxiety Risk")

plt.ylabel("Average Charging Station Distance (km)")

plt.show()
```

```python
# Range anxiety vs battery health

anxiety_battery = df.groupby(
    'range_anxiety_risk'
)['battery_health_pct'].mean().reset_index()

# Sort ascending

anxiety_battery = anxiety_battery.sort_values(
    by='battery_health_pct',
    ascending=True
)

# Ordered categories

ordered_anxiety = anxiety_battery['range_anxiety_risk']

# Plot

plt.figure(figsize=(10,5))

sns.barplot(
    data=anxiety_battery,
    x='range_anxiety_risk',
    y='battery_health_pct',
    order=ordered_anxiety
)

plt.title("Battery Health by Range Anxiety Risk")

plt.xlabel("Range Anxiety Risk")

plt.ylabel("Average Battery Health (%)")

plt.show()
```
